```python
# Initialize OpenAI chat model
chat_model = ChatOpenAI(
    model_name = "gpt-4o-mini",
    temperature = .7
)
```

```python
# manually maintaining chat history as a list
chat_history = []
```

```python
chat_history.append(SystemMessage(content="You are an expert at Apache Spark. Answer queries in a concise manner."))
```

```python
print("Chatbot: Hello! Ask me anything related to Apache Spark. Type 'exit' to quit. \n")
```

```python
while True:
  # Get the user Input
  user_input = input("You :")

  # Exit Condition
  if user_input.lower() == "exit" :
    print(chat_history)
    print("Chatbot: Goodbye! Have a great day!")
    break

  # Add user message to chat history
  chat_history.append(HumanMessage(content=user_input))

  # Invoke the model with history
  response = chat_model.invoke(chat_history)

  # print response
  print(f"Chatbot:  {response.content}")

  # Store AI response in chat history
  chat_history.append(AIMessage(content=response.content))



```
