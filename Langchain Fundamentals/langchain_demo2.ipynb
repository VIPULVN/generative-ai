```python
from langchain_community.llms import HuggingFaceHub #LangChain wrapper for hugging face models
import os
```

```python
os.environ["HUGGINGFACEHUB_API_TOKEN"] = ""
```

```python
# Inititalize the hugging face model via Langchain
chat_model = HuggingFaceHub(
    repo_id="mistralai/Mistral-7B-Instruct-v0.1",
    model_kwargs = {
        "temperature": .7, #control randomness
        "max_length": 512 #limiting the response length
    }
)
```

```python
response = chat_model.invoke("Tell me about rules of tennis game")
```
