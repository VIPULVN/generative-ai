```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o-mini", temperature = 0)
```

```python
from langchain.prompts import ChatPromptTemplate
my_prompt = ChatPromptTemplate.from_messages([
    ("system", (
        "check if the user request is a question or not"
    )),
    ("human", "User request: {question}")
])
```

```python
from operator import itemgetter

chain = itemgetter("question") | my_prompt | llm
result = chain.invoke({"question": "my name is Sumit"})
result
```

```python
from pydantic import BaseModel, Field
from typing import Literal

class QuestionChecker(BaseModel):
    is_question: Literal["yes", "no"] = Field(
        description="Specify if the input is a question. Only 'yes' or 'no' are allowed."
    )
```

```python
structured_llm = llm.with_structured_output(QuestionChecker)
```

```python
from operator import itemgetter

chain = itemgetter("question") | my_prompt | structured_llm
result = chain.invoke({"question": "My name is Sumit"})
result
```
