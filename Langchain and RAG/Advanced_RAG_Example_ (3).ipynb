```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_core.documents import Document

# Step 1: Sample long document (could be from a loader)
text = """
LangChain is a framework for developing applications powered by language models.
It enables applications that are:
- Data-aware: connect a language model to other sources of data
- Agentic: allow a language model to interact with its environment

## Features

LangChain provides modular abstractions for:
1. Prompt templates
2. LLMs
3. Agents
4. Tools
5. Memory
6. Indexing and Retrieval

## Use Cases

- Document Q&A (RAG)
- Chatbots
- Agents
- Data Augmented Generation
- Summarization
"""

# Convert into LangChain Document
doc = Document(page_content=text)
docs = [doc]
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,          # max size of each chunk
    chunk_overlap=50,        # how much overlap between chunks
    separators=["\n\n", "\n", ".", " ", ""]  # order of splitting preference
)

chunks = splitter.split_documents(docs)
```

```python
for i, chunk in enumerate(chunks, 1):
    print(f"\n--- Chunk {i} ---\n{chunk.page_content}")
```

```python
from langchain.text_splitter import CharacterTextSplitter
from langchain_core.documents import Document

# Sample long text (or use your actual document)
text = """LangChain is a framework for developing applications powered by language models.

It enables applications that are context-aware, reasoning, and modular.

This makes it easier to build agents, chatbots, RAG systems, and more using LLMs."""

# Wrap in a Document object
doc = Document(page_content=text)

# Initialize the splitter
splitter = CharacterTextSplitter(
    separator="",
    chunk_size=20,        # Max size of each chunk
    chunk_overlap=10      # Overlap between chunks
)

# Split the document
chunks = splitter.split_documents([doc])

# Display results
for i, chunk in enumerate(chunks, 1):
    print(f"\n--- Chunk {i} ---\n{chunk.page_content}")
```

```python
from langchain.text_splitter import MarkdownHeaderTextSplitter

markdown_text = """
# Introduction
LangChain helps with LLM applications...

## Use Cases
- RAG
- Agents

# Installation
Use pip install...
"""

splitter = MarkdownHeaderTextSplitter(headers_to_split_on=[("#", "Header1"), ("##", "Header2")])
docs = splitter.split_text(markdown_text)

for doc in docs:
    print(doc.metadata, doc.page_content[:100])
```
