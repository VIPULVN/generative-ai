```python
from transformers import pipeline

# pipeline() is the main factory method to load a model and its required tokenizers/processors.
# Parameter 1: "text-generation" - The unified task type replacing legacy specific tasks.
# Parameter 2: model="Qwen/Qwen2.5-1.5B-Instruct" - The exact repository ID on Hugging Face.
# We use an 'Instruct' model (like Qwen) because it is trained to follow chat commands.
summarizer = pipeline("text-generation", model="Qwen/Qwen2.5-1.5B-Instruct")

text = """Artificial Intelligence (AI) has rapidly become an integral part of our daily lives.
From voice assistants like Siri and Alexa to recommendation systems on platforms like Netflix and Amazon, AI is shaping how we interact with technology. It is no longer confined to the realm of science fiction but is now a practical tool that enhances efficiency and convenience."""

# messages: A list of dictionaries representing the conversation history.
# This chat-formatted prompt instructs the model instead of relying on pipeline task types.
# - "role": Defines who is speaking. "user" represents the human giving the instruction.
# - "content": The actual string prompt containing the instruction and the text to process.
messages = [
    {
        "role": "user",
        "content": f"Summarize the following text briefly in 1-2 sentences:\n\n{text}"
    }
]

# Inference step: Passing the formatted messages to the model.
# max_new_tokens=90: Strictly limits the model to generating a maximum of 90 new tokens (words/subwords) in its response.
# This prevents the model from rambling and acts similarly to the old max_length parameter, but only counts *new* text.
output = summarizer(messages, max_new_tokens=90)

# Decoding the nested output structure:
# output[0]: Gets the first sequence generated (since we didn't ask for multiple sequences).
# ["generated_text"]: The key where the pipeline stores the full conversation history (input + output).
# [-1]: Accesses the very last dictionary in the conversation list (which is the model's new response).
# ["content"]: Extracts the raw string text from that final response.
print(output[0]["generated_text"][-1]["content"])
print(output)
```

```python
from transformers import pipeline

# Initializing the same instruction-tuned model for text-generation[cite: 2].
qa_pipe = pipeline("text-generation", model="Qwen/Qwen2.5-1.5B-Instruct")

contxt = "The Transformers library by Hugging Face provides state-of-the-art NLP models. It supports tasks like translation, summarization, and question answering."
qus = "What does the Transformers library support?"

# messages parameter: Replaces the old `question=` and `context=` parameters.
# We manually format the string to contain both the context and the question, explicitly instructing the model to answer based *only* on the context[cite: 2].
messages = [
    {
        "role": "user",
        "content": f"Answer the question based only on the provided context.\n\nContext:{contxt}\n\nQuestion: {qus}"
    }
]

# max_new_tokens=50: Caps the model's generated answer at 50 tokens to ensure a concise response.
output = qa_pipe(messages, max_new_tokens=50)

# Extracts the final assistant reply from the chat history dictionary.
print(output[0]["generated_text"][-1]["content"])
```

```python
from transformers import pipeline

# Reusing the unified text-generation pipeline paradigm[cite: 2].
translator_pipe = pipeline("text-generation", model="Qwen/Qwen2.5-1.5B-Instruct")

text = "Trendytech offers high quality big data trainings."

# messages parameter: We inject the text variable into the content string,
# instructing the model specifically on the source and target languages (English to French)[cite: 2].
messages = [
    {
        "role": "user",
        "content": f"Translate the following English text to French:\n\n{text}"
    }
]

# max_new_tokens=50: Ensures the generation stops after 50 tokens, suitable for short sentence translation.
output = translator_pipe(messages, max_new_tokens=50)

# Parses the list-of-dictionaries output to isolate just the translated string.
print(output[0]["generated_text"][-1]["content"])
```
